# text-to-speech

> Small text-to-speech models for phones: pretrained ones under full playback control, and one trained from scratch.

## Two lines of research

**1. Pretrained models with system-voice control** (notebooks `1xx`). Run open TTS models on this Mac, then on iOS and
Android, with the same control `~/audio-reader` has over Alex (iOS) and the Android system voices: play, pause, rewind,
fast forward and change speed mid-utterance. That means streaming synthesis in small chunks, knowing which audio sample
belongs to which word, and changing rate without re-synthesising everything. The milestone order is: controllable on
the Mac, then compiled for each phone (Core ML; LiteRT / ONNX Runtime / ExecuTorch) and just as controllable there.

**2. A voice from scratch that beats Alex** (notebooks `0xx`). Build every piece ourselves, starting from raw audio, and
train a small model that sounds subjectively better than Alex while staying small and export-friendly enough for line 1's
phone runtime.

| `04_vocoder_improvements` | | More data, LR decay and a wider generator, one at a time |
| `100_pretrained_voice` | `pretrained` | Kokoro-82M checked against audio-reader's controls: speed, word timings, rate |
| `101_faster_speech` | | Naive speed-ups and six time-stretchers against Kokoro's own speed, 1.5x to 3x: Sonic wins |

| `103_hardware` | `deploy` | Kokoro's decoder on the Mac's GPU and Neural Engine via Core ML, smaller weights, XNNPACK |
| `104_phone` | `phone` | The decoder on a Pixel 10a through LiteRT: the converter's bugs, CPU against GPU |
| `999_fourier` | | The FFT from scratch |

Code lives in the notebooks. Cells marked `#| export` are written to `text_to_speech/` by `nbdev-export`, so never edit
the `.py` files directly.
